```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName('pyspark-pilo')
    .master('local[*]')
    .getOrCreate()
)
```

```python
sales_df = spark.read.parquet('/home/spark/data/sales/')
```

```python
sales_df.printSchema()
```

```python
sales_df.rdd.getNumPartitions()
```

```python
sales_df.write.mode('overwrite').parquet('/home/spark/data/test')
```

```python
from pyspark.sql.functions import date_format, col

sales_year_df = sales_df.withColumn('month', date_format('transacted_at', 'MM'))
```

```python
sales_year_df.write.partitionBy('month').mode('overwrite').parquet('/home/spark/data/test')
```
