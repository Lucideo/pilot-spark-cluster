```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName('pyspark-pylot')
    .master('local[*]')
    .getOrCreate()
)

spark
```

```python
# Reading employee csv files and defining schemas

emp_df = spark.read.schema('employee_id string, department_id string, name string, age int, gender string, salary double, hire_date date').csv('/home/spark/data/employees', header=True)

emp_df.printSchema()
```

```python
# Using .withColumn()

from pyspark.sql.functions import col

emp_tax_df = emp_df.withColumn('tax', col('salary') * 0.2)
emp_tax_df.show()
```

```python
# Using lit()

from pyspark.sql.functions import lit

emp_tax_df = emp_tax_df = emp_tax_df.withColumn('one', lit(1)).withColumn('two', lit('two'))
emp_tax_df.show()
```

```python
# Renaming columns

emp_tax_df.withColumnRenamed('tax', 'taxes').show()
```

```python
# Drop columns

emp_tax_df.drop('tax', 'one', 'two').show()
```

```python
# Adding more than one columns

columns = {
    'tax': col('salary') * 0.2,
    'one': lit(1),
    'two': lit('two')
}

emp_tax_df = emp_tax_df.withColumns(columns)
```

```python
emp_tax_df.write.mode('overwrite').csv('/home/spark/data/employees_tax', header=True)
```
