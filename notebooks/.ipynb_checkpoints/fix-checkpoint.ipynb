```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName('pyspark-pilot')
    .master('spark://spark-master:7077')
    .config('spark.executor.instances', 4)
    .config('spark.executor.cores', 2)
    .config('spark.executor.memory', '2048M')
    .getOrCreate()
)

spark
```

```python
emp_df = spark.read.csv('/home/spark/data/employee/csv', header=True)
```

```python
emp_df.printSchema()
```

```python
from pyspark.sql.functions import col

emp_new_df = emp_df.withColumn('tax', col('salary') * 0.2)
```

```python
emp_new_df.write.mode('overwrite').parquet('/home/spark/data/employee/parquet')
```

```python
emp_sales = spark.read.parquet('/home/spark/data/sales')
```

```python
emp_sales.show()
```

```python
spark.stop()
```
