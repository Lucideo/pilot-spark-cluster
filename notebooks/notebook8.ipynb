```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName('pyspark-pilot')
    .master('local[*]')
    .getOrCreate()
)
```

```python
emp_schema = "employee_id string, department_id string, name string, age integer, gender string, salary double, hire_date date, tax double"

emp_df = spark.read.schema(emp_schema).csv('/home/spark/data/employee', header=True)
# emp_df = spark.read.csv('/home/spark/data/employee', header=True)
# emp_df = spark.read.csv('/home/spark/data/employee', header=True, inferSchema=True)
```
